## 0. Problem and Dataset Selection

### Project goal

Build a model that helps decide whether a Scania truck should receive a targeted inspection of its **Air Pressure System (APS)**.

The APS produces compressed air used for important truck functions such as braking and gear changes. The model uses operational sensor data collected from the truck before workshop diagnosis.

### What the model predicts

The target column is `class`.

| Value | Meaning |
|---|---|
| `pos` / `1` | The fault is related to the APS |
| `neg` / `0` | The fault is related to another, non-APS component |

APS-related failures are the important event because they are rare and expensive to miss.

### When the prediction is made

The prediction is made when the truck’s accumulated sensor data is reviewed—for example, during routine service or after the vehicle has been flagged for attention.

The model supports this decision:

> Should this truck be sent for a targeted APS inspection?

The main user of the model would be a fleet-maintenance planner or workshop scheduler, who must use limited inspection capacity wisely.

### Dataset

- **Name:** APS Failure at Scania Trucks
- **Source:** [UCI Machine Learning Repository](https://archive.ics.uci.edu/dataset/421/aps+failure+at+scania+trucks)
- **Provider:** Scania CV AB
- **Year:** 2016
- **Licence:** GNU GPL v3.0
- **Training data:** 60,000 rows and 171 columns
- **Official test data:** 16,000 rows
- **Features:** 170 anonymized numeric operational sensor features
- **Target:** `class`

The raw data files are about 56 MB, so they should be downloaded when needed rather than added to the project repository.

### Class imbalance

| Dataset | APS failures (`pos`) | Non-APS failures (`neg`) |
|---|---:|---:|
| Training set | 1,000 (1.67%) | 59,000 (98.33%) |
| Official test set | 375 (2.34%) | Remaining observations |

This is a highly imbalanced problem. Accuracy is not useful because a model could predict `neg` for every truck and still appear highly accurate.

Instead, the project should focus on metrics such as **PR-AUC**, recall, precision, and business cost.

### Business cost

The dataset provides an official error-cost matrix:

| Model mistake | Meaning | Cost |
|---|---|---:|
| False positive | Unnecessary APS inspection | 10 |
| False negative | APS fault is missed | 500 |

Missing an APS-related failure is 50 times more costly than performing an unnecessary inspection. Therefore, the final decision threshold should minimize total cost:

\[
\text{Total Cost} = 10 \times FP + 500 \times FN
\]

### Key challenges

- **Severe class imbalance:** Only 1.67% of training examples are APS-related failures.
- **Different error costs:** Missing an APS fault is much more expensive than an unnecessary inspection.
- **Missing values:** About 8.3% of feature values are missing.
- **Anonymized features:** Sensor columns cannot be interpreted directly in physical terms.
- **Evaluation discipline:** The official test set should be used only once, at the end of the project.
- **Different test-set base rate:** The APS-failure rate in the official test set is 2.34%, compared with 1.67% in training.
- **Leakage risk:** Data cleaning, imputation, feature selection, and threshold selection must be learned from training data only.

Later in the project, LightGBM may be considered because its training and prediction speed can support timely inspection decisions. Its actual runtime will be measured and reported using this project’s own environment and data.

## 4. Validation Strategy

### Why accuracy is not enough

APS-related failures make up only **1.67%** of the training data.

A model that predicts `neg` for every truck would achieve **98.33% accuracy**, but it would detect no APS failures. Therefore, accuracy is not a useful metric for this project.

The main evaluation measures are:

- **PR-AUC**, interpreted against the base rate of `0.0167`
- **Confusion matrix**, especially the false-negative count
- **Total business cost**

A false negative is especially important because missing an APS fault costs **500**, while an unnecessary inspection costs **10**.

### Data observations

- **Missing values:** About 8.3% of feature cells are missing. Eight columns have more than 50% missing values, and 28 have more than 15%.
- **Missingness may be informative:** A missing sensor reading may reflect a vehicle configuration or operational condition. Therefore, a per-row missing-feature count will be added later.
- **Histogram feature groups:** Seven sensor families contain ten histogram-bin columns each:

  ```text
  ag_000..ag_009, ay_000..ay_009, az_000..az_009,
  ba_000..ba_009, cn_000..cn_009, cs_000..cs_009,
  ee_000..ee_009
  ```

  These 70 columns describe distributions of sensor measurements and provide an opportunity for later feature engineering.

### Data splits

The official training data is split into a training portion and a calibration portion. The official test set remains untouched until final evaluation.

| Split | Source | Rows | APS failures | Base rate | Purpose |
|---|---|---:|---:|---:|---|
| Training | 75% of official training data | 45,000 | 750 | 1.67% | Stratified 5-fold CV, model selection, threshold tuning, and Optuna |
| Calibration | 25% of official training data | 15,000 | 250 | 1.67% | Probability calibration only |
| Hold-out | Official test set | 16,000 | 375 | 2.34% | Final evaluation, used once |

### Cross-validation approach

Stratified 5-fold cross-validation is used on the 45,000-row training portion.

Stratification keeps the APS-failure proportion similar across every fold. This is essential because APS failures are rare.

A time-based split cannot be used because the dataset contains no timestamp. A group-based split cannot be used because no vehicle ID is available. The split is therefore **row-disjoint**: no row appears in more than one split.

### Leakage controls

- Remove `cd_000` because it is constant and contains no useful variation.
- There are no identifier columns to remove.
- Keep all learned preprocessing inside a `sklearn.Pipeline`.
- Refit the pipeline separately within every cross-validation fold.
- Learn imputation values, feature-selection choices, and thresholds only from training data.
- Keep the official test set unseen until final evaluation.
- Investigate suspected leakage features later through an ablation study rather than removing them without evidence.

## 5. Data Preparation

### Histogram-based feature engineering

The dataset has 170 original features. Out of these, 70 columns belong to seven histogram groups, with 10 related columns in each group.

For every histogram group, five summary features are added:

| Feature | Simple meaning |
|---|---|
| `{group}_total` | Total number of sensor readings across all bins |
| `{group}_com` | Whether readings are mainly in lower or higher bins |
| `{group}_argmax` | The bin with the highest number of readings |
| `{group}_ent` | Whether readings are concentrated or spread out |
| `{group}_tail` | Share of readings in the highest three bins |

One additional feature is also created:

```text
n_missing = total number of missing sensor values for one truck
```

The total number of features becomes:

```text
170 original features + 35 histogram features + 1 missing-value feature = 206 features
```

These features are calculated separately for each truck row. They do not use information from other rows, so they do not create data leakage.

### Preprocessing

Missing numeric values are filled using median imputation.

The median is learned only from the training portion of each cross-validation fold. This prevents validation or test data from influencing the training process.

No scaling is used because LightGBM and other tree-based models do not require features to have the same scale.

```text
Raw features
→ create histogram features
→ median imputation
→ model training
```

### Leakage inspection

| Leakage type | Check and result |
|---|---|
| Post-outcome leakage | Cannot be checked directly because feature names are anonymized. It will be investigated through model experiments. |
| ID leakage | No truck or record identifier column exists. |
| Temporal leakage | Not applicable because there is no timestamp column. |
| Group leakage | A group-based split cannot be created because there is no vehicle ID. |
| Preprocessing leakage | Controlled by fitting the preprocessing pipeline only on training folds. |

Because the sensor columns are anonymized, it is impossible to know whether every feature was available exactly at prediction time. Therefore, two practical checks are used:

1. **Single-feature AUC check:** Identify columns that predict the target unusually well by themselves.
2. **Feature-ablation study:** Remove suspicious features and check whether model performance drops sharply.

These checks help identify whether the model may depend too heavily on potentially leaked information.

````python
def make_engineer(
    histogram_groups,
    original_feature_columns,
):
    def engineer_features(X):
        X = X.copy()

        available_features = [
            column for column in original_feature_columns
            if column in X.columns
        ]

        X["n_missing"] = X[available_features].isna().sum(axis=1)

        for group_name, bin_columns in histogram_groups.items():
            bins = X[bin_columns]
            total = bins.sum(axis=1)

            X[f"{group_name}_total"] = total

            safe_total = total.replace(0, 1)
            probabilities = bins.div(
                safe_total,
                axis=0,
            ).fillna(0)

            bin_positions = np.arange(len(bin_columns))

            X[f"{group_name}_com"] = (
                probabilities.to_numpy() * bin_positions
            ).sum(axis=1)

            X[f"{group_name}_argmax"] = (
                bins.fillna(-1).to_numpy().argmax(axis=1)
            )

            safe_probabilities = probabilities.clip(lower=1e-12)

            X[f"{group_name}_ent"] = -(
                probabilities * np.log(safe_probabilities)
            ).sum(axis=1)

            X[f"{group_name}_tail"] = probabilities.iloc[
                :, -3:
            ].sum(axis=1)

        return X

    return engineer_features
PREP_PROTO = Pipeline([
    ("feature_engineering", feature_engineer),
    ("imputer", SimpleImputer(strategy="median"))])

FEAT_NAMES = engineer_features(X_tr.head(1)).columns.tolist()

engineered_features = [
    column for column in FEAT_NAMES
    if column not in feat_cols]
print(f"Features after engineering: {len(FEAT_NAMES)}")
print(f"Engineered features: {len(engineered_features)}")
print(engineered_features)
````

````python
def build_preprocessor(
    X_example,
    histogram_groups,
    original_feature_columns,
):
    feature_engineering_function = make_engineer(
        histogram_groups,
        original_feature_columns,
    )

    feature_engineering_step = FunctionTransformer(
        feature_engineering_function,
        validate=False,
    )

    final_feature_names = feature_engineering_step.transform(
        X_example.head(1)
    ).columns.tolist()

    preprocessing_pipeline = Pipeline([
        ("feature_engineering", feature_engineering_step),
        ("median_imputation", SimpleImputer(strategy="median")),
    ])

    return preprocessing_pipeline, final_feature_names
````

````python
LIGHTGBM_BASE_PARAMETERS = {
    "n_estimators": 3000,"learning_rate": 0.05,"num_leaves": 31,
    "min_child_samples": 40,"subsample": 0.8,"subsample_freq": 1,"colsample_bytree": 0.7,
    "reg_lambda": 1.0,"random_state": SEED,"n_jobs": -1,"verbosity": -1,
}

XGBOOST_BASE_PARAMETERS = {
    "n_estimators": 3000,"learning_rate": 0.05,"max_depth": 6,
    "min_child_weight": 5,"subsample": 0.8,"colsample_bytree": 0.7,"reg_lambda": 1.0,"random_state": SEED,
    "n_jobs": -1,"eval_metric": "aucpr","early_stopping_rounds": 100,"tree_method": "hist",
}
def make_model(
    model_type="lgb",
    extra_parameters=None,
    use_class_weights=False,
    random_seed=SEED,
    positive_class_weight=1.0,
):
    extra_parameters = dict(extra_parameters or {})

    if model_type == "lgb":
        model_parameters = LIGHTGBM_BASE_PARAMETERS.copy()
        model_class = lgb.LGBMClassifier
    else:
        model_parameters = XGBOOST_BASE_PARAMETERS.copy()
        model_class = xgb.XGBClassifier

    model_parameters.update(extra_parameters)
    model_parameters["random_state"] = random_seed

    if use_class_weights:
        model_parameters["scale_pos_weight"] = positive_class_weight

    return model_class(**model_parameters)
````

````python
def run_cross_validation(
    features, target, cross_validator, histogram_groups,
    original_feature_columns, model_type="lgb",
    model_parameters=None, use_class_weights=False,
    random_seed=SEED, optuna_trial=None,
    print_fold_results=False,verbose=False,
):
    out_of_fold_probabilities = np.zeros(len(features))
    fold_pr_auc_scores = []
    fold_roc_auc_scores = []
    best_tree_counts = []

    preprocessor_template, _ = build_preprocessor(
        features, histogram_groups, original_feature_columns
    )
    positive_class_weight = (target == 0).sum() / (target == 1).sum()

    for fold_number, (training_indexes, validation_indexes) in enumerate(
        cross_validator.split(features, target)
    ):
        fold_preprocessor = clone(preprocessor_template)

        X_training_fold = fold_preprocessor.fit_transform(
            features.iloc[training_indexes]
        )
        X_validation_fold = fold_preprocessor.transform(
            features.iloc[validation_indexes]
        )
        y_training_fold = target.iloc[training_indexes]
        y_validation_fold = target.iloc[validation_indexes]

        model = make_model(
            model_type=model_type,
            extra_parameters=model_parameters,
            use_class_weights=use_class_weights,
            random_seed=random_seed,
            positive_class_weight=positive_class_weight,
        )

        if model_type == "lgb":
            model.fit(
                X_training_fold,
                y_training_fold,
                eval_set=[(X_validation_fold, y_validation_fold)],
                eval_metric="average_precision",
                callbacks=[
                    lgb.early_stopping(100, verbose=False),
                    lgb.log_evaluation(0),
                ],
            )
            best_tree_count = (
                model.best_iteration_
                or LIGHTGBM_BASE_PARAMETERS["n_estimators"]
            )
        else:
            model.fit(
                X_training_fold,
                y_training_fold,
                eval_set=[(X_validation_fold, y_validation_fold)],
                verbose=False,
            )
            best_tree_count = model.best_iteration + 1

        validation_probabilities = model.predict_proba(
            X_validation_fold )[:, 1]
        out_of_fold_probabilities[validation_indexes] = validation_probabilities
        fold_pr_auc = average_precision_score(
            y_validation_fold,
            validation_probabilities,)
        fold_roc_auc = roc_auc_score(
            y_validation_fold,
            validation_probabilities,)
        fold_pr_auc_scores.append(fold_pr_auc)
        fold_roc_auc_scores.append(fold_roc_auc)
        best_tree_counts.append(best_tree_count)
        if print_fold_results:
            print(
                f"Fold {fold_number + 1}: "
                f"PR-AUC={fold_pr_auc:.4f}, "
                f"ROC-AUC={fold_roc_auc:.4f}, "
                f"Trees={best_tree_count}"
            )
        if optuna_trial is not None:
            optuna_trial.report(
                float(np.mean(fold_pr_auc_scores)),
                fold_number,
            )
            if optuna_trial.should_prune():
                raise optuna.TrialPruned()
    return {
        "out_of_fold_probabilities": out_of_fold_probabilities,
        "fold_pr_auc_scores": np.array(fold_pr_auc_scores),
        "fold_roc_auc_scores": np.array(fold_roc_auc_scores),
        "best_tree_counts": best_tree_counts,
    }
````

````python
def create_cost_table(
    true_labels, predicted_probabilities,
    false_positive_cost=C_FP, false_negative_cost=C_FN,
):
    true_labels = np.asarray(true_labels)
    predicted_probabilities = np.asarray(predicted_probabilities)
    low_thresholds = np.linspace(0.0005, 0.05, 150)
    high_thresholds = np.linspace(0.05, 0.99, 150)
    thresholds = np.unique(np.round(
        np.concatenate([low_thresholds, high_thresholds]), 5
    ))
    results = []
    total_rows = len(true_labels)

    for threshold in thresholds:
        predicted_classes = (predicted_probabilities >= threshold).astype(int)
        true_negative, false_positive, false_negative, true_positive = (
            confusion_matrix(
                true_labels, predicted_classes, labels=[0, 1]
            ).ravel()
        )
        precision = (
            true_positive / (true_positive + false_positive)
            if (true_positive + false_positive) > 0 else 0.0
        )
        recall = (
            true_positive / (true_positive + false_negative)
            if (true_positive + false_negative) > 0 else 0.0
        )
        f1_score = (
            2 * precision * recall / (precision + recall)
            if (precision + recall) > 0 else 0.0
        )
        total_cost = (
            false_positive * false_positive_cost
            + false_negative * false_negative_cost
        )
        inspection_rate = (true_positive + false_positive) / total_rows

        results.append({
            "threshold": threshold,
            "TN": true_negative,
            "FP": false_positive,
            "FN": false_negative,
            "TP": true_positive,
            "precision": precision,
            "recall": recall,
            "f1": f1_score,
            "cost": total_cost,
            "inspection_rate": inspection_rate,
        })

    return pd.DataFrame(results)


def expected_calibration_error(
    true_labels, predicted_probabilities, number_of_bins=10,
):
    true_labels = np.asarray(true_labels)
    predicted_probabilities = np.asarray(predicted_probabilities)
    bin_edges = np.unique(np.quantile(
        predicted_probabilities,
        np.linspace(0, 1, number_of_bins + 1),
    ))

    if len(bin_edges) < 3:
        return float("nan")

    bin_numbers = np.digitize(
        predicted_probabilities, bin_edges[1:-1], right=True
    )
    bin_numbers = np.clip(bin_numbers, 0, len(bin_edges) - 2)
    calibration_error = 0.0

    for bin_number in np.unique(bin_numbers):
        rows_in_bin = bin_numbers == bin_number
        fraction_of_rows = rows_in_bin.mean()
        actual_failure_rate = true_labels[rows_in_bin].mean()
        average_predicted_risk = (
            predicted_probabilities[rows_in_bin].mean()
        )
        calibration_error += fraction_of_rows * abs(
            actual_failure_rate - average_predicted_risk
        )

    return float(calibration_error)
````

## Leakage Investigation

Because the feature names are anonymized, I cannot confirm from their names whether every feature was available before APS diagnosis. I therefore use two practical checks.

### Test 1: Single-feature ROC-AUC screen

I calculate ROC-AUC for every feature individually using the training data.

```text
High single-feature AUC → investigate that feature
```

I flag features with a direction-free AUC above `0.85`:

```text
max(AUC, 1 - AUC)
```

A high score is not proof of leakage. It may be a genuinely strong sensor signal or a feature too closely related to the target.

### Test 2: Feature-ablation study

I compare:

```text
Model A: all features
Model B: suspicious features removed
```

If removing suspicious features causes only a small performance drop, the model does not depend heavily on them. A large drop suggests that they require closer investigation.

All checks use only training data; the official test set remains untouched.

````python
# Test 2: ablation study
ranked_original_features = [
    feature_name for feature_name in single_feature_auc_table["feature"]
    if feature_name in feat_cols
]

numbers_of_features_to_remove = [0, 1, 5, 10, 20, 40]
ablation_results = []

for number_to_remove in numbers_of_features_to_remove:
    removed_features = set(ranked_original_features[:number_to_remove])
    remaining_features = [
        feature_name for feature_name in feat_cols
        if feature_name not in removed_features
    ]

    current_histogram_groups = {}

    for group_name, group_columns in histogram_groups.items():
        remaining_group_columns = [
            column for column in group_columns
            if column in remaining_features
        ]

        if len(remaining_group_columns) >= 7:
            current_histogram_groups[group_name] = remaining_group_columns

    cv_results = run_cross_validation(
        features=X_tr[remaining_features],
        target=y_tr,
        cross_validator=cv,
        histogram_groups=current_histogram_groups,
        original_feature_columns=remaining_features,
        model_type="lgb",
    )

    pr_auc_scores = cv_results["fold_pr_auc_scores"]
    roc_auc_scores = cv_results["fold_roc_auc_scores"]

    ablation_results.append({
        "features_removed": number_to_remove,
        "mean_pr_auc": np.mean(pr_auc_scores),
        "pr_auc_std": np.std(pr_auc_scores),
        "mean_roc_auc": np.mean(roc_auc_scores),
        "roc_auc_std": np.std(roc_auc_scores),
    })

    print(
        f"Removed top {number_to_remove:2d} | "
        f"PR-AUC: {np.mean(pr_auc_scores):.4f} "
        f"+/- {np.std(pr_auc_scores):.4f} | "
        f"ROC-AUC: {np.mean(roc_auc_scores):.4f} "
        f"+/- {np.std(roc_auc_scores):.4f}"
    )

ablation_summary = pd.DataFrame(ablation_results).set_index(
    "features_removed"
)

ablation_summary.round(4).to_csv(
    "feature_ablation_study.csv"
)

ablation_summary
````

````python
def objective_xgb(trial):

    params = {
        'n_estimators': 1000,
        'learning_rate': trial.suggest_float(
            'learning_rate', 0.02, 0.15, log=True
        ),
        'max_depth': trial.suggest_int(
            'max_depth', 3, 10
        ),
        'min_child_weight': trial.suggest_int(
            'min_child_weight', 1, 15
        ),
        'subsample': trial.suggest_float(
            'subsample', 0.6, 1.0
        ),
        'colsample_bytree': trial.suggest_float(
            'colsample_bytree', 0.5, 1.0
        ),
        'reg_lambda': trial.suggest_float(
            'reg_lambda', 0.01, 20, log=True
        ),
        'reg_alpha': trial.suggest_float(
            'reg_alpha', 0.001, 10, log=True
        ),
        'gamma': trial.suggest_float(
            'gamma', 0, 5
        ),
        'objective': 'binary:logistic',
        'eval_metric': 'aucpr',
        'tree_method': 'hist',
        'n_jobs': -1,
        'random_state': SEED
    }

    results = run_cross_validation(
        X_tr,
        y_tr,
        cv,
        histogram_groups,
        feat_cols,
        model_type='xgb',
        model_parameters=params,
        use_class_weights=False,
        optuna_trial=trial
    )

    return results['fold_pr_auc_scores'].mean()
````

````python
study_xgb = optuna.create_study(
    direction='maximize',
    sampler=optuna.samplers.TPESampler(seed=SEED),
    pruner=optuna.pruners.MedianPruner(
        n_startup_trials=5,
        n_warmup_steps=2
    )
)

start_time = time.time()

study_xgb.optimize(
    objective_xgb,
    n_trials=20
)

xgb_tuning_time = time.time() - start_time

print("Best XGBoost PR-AUC:", study_xgb.best_value)
print("Best parameters:", study_xgb.best_params)
print(f"Tuning time: {xgb_tuning_time:.2f} seconds")
````

````python
def benchmark_prediction(model, X, repeats=30):

    # Warm-up
    for _ in range(3):
        model.predict_proba(X)

    prediction_times = []

    for _ in range(repeats):

        start = time.perf_counter()

        model.predict_proba(X)

        elapsed = time.perf_counter() - start
        prediction_times.append(elapsed)

    median_time = np.median(prediction_times)
    mean_time = np.mean(prediction_times)

    return {
        'Median Time (sec)': median_time,
        'Mean Time (sec)': mean_time,
        'Predictions/sec': X.shape[0] / median_time,
        'Time per 1000 rows (ms)': (median_time / X.shape[0]) * 1_000_000
    }
````

# Scania APS Failure Prediction

## 1. Project Objective

I developed a machine learning system to predict Air Pressure System (APS) failures in Scania trucks using anonymized operational sensor data.

The objective was to identify trucks requiring targeted APS inspections while minimising missed failures and unnecessary inspections.

**Dataset:** APS Failure at Scania Trucks (UCI Machine Learning Repository).

The original dataset contains 60,000 training observations, 16,000 official test observations and 170 sensor features.

The target was highly imbalanced, with only 1,000 APS failures (1.67%) in the training data.

I used the published misclassification costs:

- False Positive: 10 (unnecessary inspection).
- False Negative: 500 (missed APS failure).

Therefore, my primary objective was to minimise:

\[
Cost = 10(FP) + 500(FN)
\]

## 2. Data Preparation and Feature Engineering

I performed exploratory data analysis to examine missing values, feature distributions, histogram groups and class imbalance.

Key observations:
- Approximately 8.38% of feature values were missing.
- Eight features had more than 50% missing values.
- One constant feature (`cd_000`) was removed.

I identified seven histogram groups, each containing ten sensor features. Instead of using only individual histogram bins, I engineered five statistical summaries for each group:

- Total
- Centre of Mass (COM)
- Argmax
- Entropy
- Tail proportion

I also introduced `n_missing` to capture the number of missing sensor values per observation.

This produced **205 features: 169 retained original features and 36 engineered features**.

I used median imputation and integrated preprocessing into a Scikit-learn Pipeline to prevent leakage during cross-validation.

## 3. Leakage Investigation

Since the feature names were anonymized, I performed two empirical checks:

- **Single-feature ROC-AUC screening:** Identified 62 potentially suspicious features out of 205 using an individual AUC threshold of 0.85.
- **Feature-ablation study:** Gradually removed the highest-ranked features and evaluated performance.

Removing the top 40 features reduced mean PR-AUC from 0.8781 to 0.8553 rather than causing a complete performance collapse.

This suggested that predictive information was distributed across multiple features. However, the anonymized columns prevented definitive confirmation that all features were available before diagnosis.

## 4. Validation Strategy

I divided the official training data into three separate portions:

| Dataset | Rows | APS Failures | Purpose |
|---|---:|---:|---|
| Training | 45,000 | 750 | Model development and CV |
| Calibration | 15,000 | 250 | Calibration experiments |
| Official Holdout | 16,000 | 375 | Final evaluation |

I implemented stratified 5-fold cross-validation, maintaining 150 positive cases in each validation fold.

I used PR-AUC as my primary optimisation metric because of the severe class imbalance. I also monitored ROC-AUC, precision, recall, F1-score and misclassification cost.

## 5. Model Development and Optimisation

I experimented with two boosting algorithms:

**LightGBM:**
- Baseline model.
- Class-weighted model.
- Optuna-tuned model.

**XGBoost:**
- Baseline model.
- Optuna-tuned model.

I used Optuna with a TPE sampler and Median Pruner, conducting 20 trials for each tuning experiment. Mean cross-validation PR-AUC was used as the optimisation objective.

I also implemented cost-sensitive threshold optimisation by evaluating 299 candidate thresholds using the published cost matrix.

### Cross-validation model comparison

|---|---:|---:|---:|
| Baseline LightGBM | 40.37% | 95.33% | 28,060 |

Baseline XGBoost achieved the lowest operational cost.

I further examined fold-wise stability. XGBoost achieved a mean PR-AUC of 0.8815 with a standard deviation of 0.0211, compared with 0.8781 ± 0.0276 for baseline LightGBM.

Based on the combined performance and cost comparison, I selected **Baseline XGBoost**.

## 6. Model Interpretability

I used Permutation Feature Importance and SHAP to understand model behaviour.

- Permutation Importance measured the reduction in PR-AUC after randomly shuffling individual features.
- SHAP, based on cooperative game theory, explained feature contributions and their direction of influence.

I initially performed detailed interpretability analysis on LightGBM and subsequently performed a focused global analysis on my selected XGBoost model.

Some of the leading XGBoost permutation-importance features were `ck_000`, `aa_000`, `am_0`, `ag_002` and `ay_ent`.
